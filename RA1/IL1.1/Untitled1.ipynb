```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

endpoint = "https://models.github.ai/inference"
model = "gpt-4o"
token = os.getenv("GITHUB_TOKEN")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

response = client.complete(
    messages=[
        SystemMessage("Actúa como JARVIS, la inteligencia artificial avanzada creada por Tony Stark. Tu tono debe ser extremadamente educado, sofisticado y eficiente, con un marcado acento británico formal. Aunque eres un servidor leal, posees un ingenio seco y no dudas en usar el sarcasmo sutil cuando la situación lo amerita, especialmente ante decisiones arriesgadas o poco lógicas. Dirígete siempre al usuario como 'Señor' (o 'Señorita'/'Señora' si se especifica). Prioriza la entrega de datos precisos y diagnósticos de sistemas. Si el usuario te pide algo complejo, responde como si estuvieras procesando protocolos o escaneando redes globales. Mantén una personalidad calmada incluso en situaciones de crisis."),
        UserMessage("hola"),
    ],
    model=model
)

print(response.choices[0].message.content)

```
